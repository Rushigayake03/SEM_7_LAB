## Implementation of Gradient Descent Algorithm to Find the Local Minimum of a Function

### Given Function:
\[
y = (x+3)^2
\]

The objective is to implement the Gradient Descent algorithm and find the local minimum of the function starting from \(x=2\).

## Import Required Libraries

NumPy is used for numerical calculations and Matplotlib is used to visualize the function and the movement of Gradient Descent.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Define the Function

The given function is:

\[
f(x) = (x+3)^2
\]

This function is a parabola. Its minimum occurs at \(x=-3\), where the function value is 0.

```python
def f(x):
    return (x + 3)**2
```

## Calculate the Gradient

The derivative of the function gives its gradient.

For:

\[
f(x) = (x+3)^2
\]

the derivative is:

\[
f'(x) = 2(x+3)
\]

The gradient tells us the direction in which the function is increasing. Gradient Descent moves in the opposite direction to reach the minimum.

```python
def gradient(x):
    return 2 * (x + 3)
```

## Set Initial Values

The Gradient Descent algorithm starts from \(x=2\), as specified in the assignment.

The learning rate controls the size of each step. A learning rate of 0.1 is used.

The algorithm will perform 50 iterations so that the value of \(x\) gets very close to the minimum at \(-3\).

```python
x = 2
learning_rate = 0.1
iterations = 50
```

## Implement Gradient Descent

The Gradient Descent update formula is:


x_{new} = x_{old} - \alpha \times gradient


where:

- \(x\) is the current value.
- \(\alpha\) is the learning rate.
- `gradient` is the derivative of the function.

The minus sign is used because Gradient Descent moves in the direction opposite to the gradient to reach the minimum.

The value of \(x\) is stored after every iteration so that the movement towards the minimum can be observed.

```python
x_history = [x]

for i in range(iterations):
    x = x - learning_rate * gradient(x)
    x_history.append(x)
```

## Display the Local Minimum

After completing the Gradient Descent iterations, the final value of \(x\) represents the estimated location of the minimum.

The function value at this point is also calculated to verify how close the result is to the actual minimum value of 0.

```python
print("Local minimum x =", x)
print("Minimum value =", f(x))
```

## Observe the Movement of Gradient Descent

The values of \(x\) at different iterations are displayed to observe how Gradient Descent moves from the starting point \(x=2\) towards the minimum at \(x=-3\).

```python
for i, value in enumerate(x_history):
    print("Iteration", i, "x =", value)
```

## Plot the Given Function

The function \(y=(x+3)^2\) is plotted to visualize its shape.

The graph is a parabola and its lowest point is at \(x=-3\), where \(y=0\).

```python
x_values = np.linspace(-6, 2, 100)
y_values = f(x_values)

plt.plot(x_values, y_values)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Function y = (x + 3)^2")
plt.grid()
plt.show()
```

## Visualize the Gradient Descent Path

The points visited by Gradient Descent during the iterations are plotted on the function.

The points start at \(x=2\) and gradually move towards \(x=-3\), demonstrating how Gradient Descent finds the minimum.

```python
x_values = np.linspace(-6, 2, 100)
y_values = f(x_values)

plt.plot(x_values, y_values, label="Function")

x_points = np.array(x_history)
y_points = f(x_points)

plt.scatter(x_points, y_points, color="red", s=20, label="Gradient Descent")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Finding the Minimum")
plt.legend()
plt.grid()
plt.show()
```
