## Implementation of K-Nearest Neighbors (K-NN) Algorithm on Diabetes Dataset

The objective is to implement the K-Nearest Neighbors classification algorithm on the `diabetes.csv` dataset and calculate:

- Confusion Matrix
- Accuracy
- Error Rate
- Precision
- Recall

The model predicts whether a patient has diabetes or not using medical predictor variables.

## Import Required Libraries

We import the libraries required for data handling, preprocessing, machine learning, and evaluation.

- `pandas` is used to read and manipulate the dataset.
- `numpy` is used for numerical operations.
- `matplotlib` is used for plotting.
- `seaborn` is used to display the confusion matrix as a heatmap.
- `train_test_split` divides the dataset into training and testing data.
- `StandardScaler` scales the features.
- `KNeighborsClassifier` implements the K-NN algorithm.
- Evaluation functions are used to calculate accuracy, precision, recall, and the confusion matrix.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    classification_report
)
```

## Load the Diabetes Dataset

The `diabetes.csv` file contains medical information about patients.

The dataset contains predictor variables such as pregnancies, glucose level, blood pressure, BMI, insulin, age, etc.

The `Outcome` column is the target variable:

- `0` = Patient does not have diabetes
- `1` = Patient has diabetes

We use `read_csv()` to load the CSV file into a Pandas DataFrame.

```python
df = pd.read_csv(r"C:\Users\rushi\Downloads\archive (1)\diabetes.csv")
df.head()
```

## Check the Dataset Information

Before building the model, we inspect the structure of the dataset.

`df.info()` shows:

- Number of rows
- Number of columns
- Column names
- Data types
- Number of non-null values

This helps us understand the dataset before preprocessing.

```python
df.info()
```

## Check for Missing Values

Missing values can cause problems while training a machine learning model.

`isnull()` checks whether values are missing.

`sum()` counts the number of missing values in each column.

If the result is zero for every column, there are no missing values that need to be handled.

```python
df.isnull().sum()
```

## Separate Features and Target

Machine learning requires us to separate the input variables from the output variable.

The `Outcome` column is the target because it tells us whether the patient has diabetes or not.

All other columns are used as input features.

Therefore:

- `X` contains the input features.
- `y` contains the target variable.

```python
X = df.drop("Outcome", axis=1)
y = df["Outcome"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

## Split the Dataset into Training and Testing Data

The dataset is divided into two parts:

- Training data is used to train the K-NN model.
- Testing data is used to evaluate how well the trained model performs on unseen data.

We use 70% of the data for training and 30% for testing.

`random_state=42` makes the split reproducible, meaning we get the same split each time the code is executed.

`stratify=y` keeps the proportion of the two target classes similar in both training and testing data.

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.30,random_state=42,stratify=y )

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## Scale the Features

K-NN is a distance-based algorithm. It calculates the distance between data points to find their nearest neighbors.

Different features can have very different numerical ranges. For example, Age may have values around 20–80, while Insulin can have much larger values.

If the features are not scaled, features with larger numerical values can have too much influence on the distance calculation.

`StandardScaler` converts the features to a common scale.

We fit the scaler only on the training data and then use the same scaler to transform both training and testing data. This prevents information from the test data from influencing the training process.

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Create the K-Nearest Neighbors Model

K-Nearest Neighbors is a supervised classification algorithm.

For a new patient, K-NN finds the K training samples that are closest to that patient.

Here we use:

\[
K = 7
\]

The class occurring most frequently among these 7 nearest neighbors is used as the predicted class.

Therefore, if most of the 7 nearest patients have `Outcome = 1`, the new patient is predicted to have diabetes.

```python
knn = KNeighborsClassifier(n_neighbors=7)
```

## Train the K-NN Model

The `fit()` function trains the model using the training data.

For K-NN, training mainly involves storing the training samples and their corresponding class labels.

The model will later use these samples to find the nearest neighbors of a test patient.

```python
knn.fit(X_train, y_train)
```

## Predict the Test Data

After training the model, we use the `predict()` function to predict the outcome for each patient in the testing dataset.

The predicted values are stored in `y_pred`.

`y_test` contains the actual answers, while `y_pred` contains the model's predictions.

These two values will be compared to evaluate the model.

```python
y_pred = knn.predict(X_test)

print("Actual values:")
print(y_test.values[:10])

print("\nPredicted values:")
print(y_pred[:10])
```

## Calculate the Confusion Matrix

A confusion matrix compares the actual class with the predicted class.

For binary classification, it contains four values:

- True Negative (TN): Actual 0 and predicted 0.
- False Positive (FP): Actual 0 but predicted 1.
- False Negative (FN): Actual 1 but predicted 0.
- True Positive (TP): Actual 1 and predicted 1.

The confusion matrix helps us understand not only how many predictions were correct, but also what type of mistakes the model made.

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

## Calculate Accuracy, Error Rate, Precision and Recall

After making predictions, we evaluate the performance of the K-NN model using four important measures.

### 1. Accuracy

Accuracy represents the percentage of total predictions that were correct.

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

### 2. Error Rate

Error rate represents the percentage of predictions that were incorrect.

$$
Error\ Rate = 1 - Accuracy
$$

### 3. Precision

Precision tells us how many of the patients predicted as diabetic were actually diabetic.

$$
Precision = \frac{TP}{TP + FP}
$$

### 4. Recall

Recall tells us how many of the patients who actually have diabetes were correctly identified by the model.

$$
Recall = \frac{TP}{TP + FN}
$$

Here, `Outcome = 1` is considered the positive class.

```python
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)

print("\nError Rate:", error_rate)
print("Error Rate (%):", error_rate * 100)

print("\nPrecision:", precision)
print("Precision (%):", precision * 100)

print("\nRecall:", recall)
print("Recall (%):", recall * 100)
```

## Classification Report

The classification report provides precision, recall, F1-score, and support for both classes.

It provides a detailed summary of the classification performance.

The assignment specifically requires precision and recall, while the classification report also provides F1-score and support.

```python
print(classification_report(y_test, y_pred))
```
